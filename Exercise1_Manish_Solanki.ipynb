```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create a Pandas DataFrame with four columns (f1, f2, f3, f4) of random numerical values
np.random.seed(42)  # Ensures reproducible random numbers
data = np.random.uniform(low=10.0, high=100.0, size=(50, 4))
df = pd.DataFrame(data, columns=['f1', 'f2', 'f3', 'f4'])

print("Initial DataFrame (first 5 rows):")
display(df.head())

# 2. Generate a scatter plot between column f2 and column f3
plt.figure(figsize=(8, 5))
plt.scatter(df['f2'], df['f3'], color='teal', alpha=0.75, edgecolors='black')
plt.title('Scatter Plot: Relationship Between f2 and f3', fontsize=14)
plt.xlabel('Column f2', fontsize=12)
plt.ylabel('Column f3', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# 3. Compute the row-wise arithmetic mean across f1, f2, f3, and f4 and store in f5
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1)

print("\nUpdated DataFrame with Row-wise Mean (f5):")
display(df.head())

# 4. Export the complete DataFrame (f1 through f5) to a Microsoft Excel file (.xlsx)
output_filename = 'Exercise1_Output.xlsx'
df.to_excel(output_filename, index=False, engine='openpyxl')
print(f"\nDataFrame successfully exported to '{output_filename}'.")
```
