```python
import gradio as gr

# 1. Define the chatbot response logic
def got_a_assistant(message, history):
    """
    Simple rule-based conversational handler for the GOT-A prototype.
    Can be swapped with an LLM API call (e.g., Gemini, OpenAI, or Hugging Face).
    """
    user_input = message.lower().strip()

    if "pandas" in user_input:
        return (
            "Pandas is a Python library for data manipulation and analysis. "
            "In Exercise 1, we used it to create DataFrames, compute row-wise means, "
            "and export results to Excel."
        )
    elif "opencv" in user_input or "blend" in user_input:
        return (
            "OpenCV (cv2) is used for computer vision tasks. "
            "The cv2.addWeighted() function blends two equally sized images using "
            "alpha and beta weights."
        )
    elif "gradio" in user_input:
        return (
            "Gradio lets you build interactive web UIs for machine learning models "
            "and chatbots directly from Python in just a few lines of code."
        )
    else:
        return (
            f"You asked: '{message}'. "
            "I am your NIELIT GOT-A AI Assistant. Try asking me about Pandas, OpenCV, or Gradio!"
        )

# 2. Configure the Gradio ChatInterface with title, description, and sample prompts
demo = gr.ChatInterface(
    fn=got_a_assistant,
    title="NIELIT GOT-A Interactive AI Assistant",
    description=(
        "Welcome to the Government Officer Training (GOT-A) AI Chatbot prototype. "
        "Type a message below or click one of the sample prompts to test the interface."
    ),
    examples=[
        "How is Pandas used in data analysis?",
        "Explain image blending in OpenCV.",
        "What is Gradio used for?"
    ]
)

# 3. Launch the web interface
# Use share=False for local execution; set inline=True to display directly inside Jupyter
demo.launch(inline=True)
```
