```python
from sklearn.linear_model import LinearRegression
# 1. Input Data (Must be 2D array for sklearn, Target is 1D)
X_pages = [[10], [20], [30], [40]]  # Page counts
y_days = [2, 4, 6, 8]             # Actual processing days

# 2. Train the Model
model = LinearRegression()
model.fit(X_pages, y_days)

# 3. Predict for a new file with 50 pages
new_file = [[50]]
predicted_days = model.predict(new_file)

print(f"Predicted days for 50 pages: {predicted_days[0]:.1f} days")
```

```python
import matplotlib.pyplot as plt
# Extract numerical values from X_pages list of lists for plotting
x_values = [x[0] for x in X_pages]

# Create the scatter plot for the actual training data
plt.scatter(x_values, y_days, color='blue', label='Actual Training Data')

# Create a line plot representing the model's learned regression line
# We'll plot the line using the original x_values and their predicted y_values
predicted_y_values = model.predict(X_pages)
plt.plot(x_values, predicted_y_values, color='red', label='Regression Line')

# Plot the new prediction point
plt.scatter(new_file[0][0], predicted_days[0], color='green', marker='*', s=100, label=f'Prediction (50 pages, {predicted_days[0]:.1f} days)')

# Add context: labels and title
plt.xlabel('Page Counts')
plt.ylabel('Processing Days')
plt.title('Linear Regression: Processing Days vs. Page Counts')

# Display the legend to differentiate the points and line
plt.legend()

# Show the plot
plt.grid(True) # Adding a grid for better readability
plt.show()
```
