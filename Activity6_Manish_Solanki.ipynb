```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

# 1. Input Features: [Income_Lakhs, Docs_Complete]
X_apps = [[2.5, 1], [8.0, 0], [3.0, 1], [12.0, 0]]
# Target Labels: 1 = Approved, 0 = Rejected
y_decision = [1, 0, 1, 0]

# 2. Train the Classifier
classifier = LogisticRegression()
classifier.fit(X_apps, y_decision)

# 3. Predict for a new applicant: ₹3.5 Lakhs income, complete documents
new_applicant = [[3.5, 1]]
result = classifier.predict(new_applicant)

status = "Approved (1)" if result[0] == 1 else "Rejected (0)"
print(f"Application Result: {status}")

# --- Matplotlib Plotting ---

# Separate the training data based on the decision for coloring
approved_incomes = [X_apps[i][0] for i in range(len(y_decision)) if y_decision[i] == 1]
approved_docs = [X_apps[i][1] for i in range(len(y_decision)) if y_decision[i] == 1]

rejected_incomes = [X_apps[i][0] for i in range(len(y_decision)) if y_decision[i] == 0]
rejected_docs = [X_apps[i][1] for i in range(len(y_decision)) if y_decision[i] == 0]

# Create the scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(approved_incomes, approved_docs, color='green', label='Approved (1)', s=100)
plt.scatter(rejected_incomes, rejected_docs, color='red', label='Rejected (0)', s=100)

# Plot the new prediction point
plt.scatter(new_applicant[0][0], new_applicant[0][1], color='blue', marker='*', s=200, label=f'New Applicant: {status}')

# Add context: labels, title, and y-axis ticks
plt.xlabel('Income (Lakhs)')
plt.ylabel('Documents Complete (0 = No, 1 = Yes)')
plt.title('Supervised Classification: Applicant Approval')
plt.yticks([0, 1]) 
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# Show the plot
plt.show()
```
